### Analisis Latihan 3:

1. Penjelasan Script:
   Secara keseluruhan, script di atas dirancang untuk menampilkan tiga baris teks informasi mengenai profil seorang tokoh bernama Budi.         Setiap baris menggunakan fungsi bawaan `print()` yang diisi dengan argumen berupa tipe data teks (*string*). Di akhir setiap baris          instruksi, terdapat komentar segaris (*inline comment*) yang diawali dengan tanda pagar (`#`). Komentar ini berfungsi sebagai               dokumentasi penjelasan kode yang akan diabaikan oleh interpreter Python saat proses eksekusi berlangsung.
2. Tindakan Perbaikan Kesalahan:
   * Pada baris pertama, fungsi `print("Ini Budi")` berjalan normal menggunakan sepasang tanda petik dua (`""`). 
   * Pada baris kedua, terjadi perbaikan pada sintaks string. Berdasarkan aturan dasar Python, penulisan teks (string) dapat diapit oleh       tanda petik dua (`""`) maupun tanda petik satu (`''`) asalkan penggunaannya konsisten dan berpasangan. Kode asli mengalami                  *SyntaxError*    karena hanya memiliki tanda petik pembuka (`'`) tanpa adanya tanda petik penutup. Perbaikan dilakukan dengan               menambahkan tanda petik satu (`'`) di akhir kalimat teks agar string menjadi valid.
   * Pada baris ketiga, terjadi perbaikan pada penulisan fungsi bawaan. Kode asli menuliskan `prnt`, yang menyebabkan kesalahan                  *NameError* karena kata kunci tersebut tidak dikenali oleh sistem. Hal ini membuktikan sifat Python yang *case-sensitive* dan               ketat terhadap penulisan instruksi. Perbaikan dilakukan dengan mengubah kata tersebut menjadi `print()`. 
3. Penjelasan Hasil:
   Setelah seluruh kesalahan sintaks di atas diperbaiki, script berhasil dieksekusi secara berurutan dari atas ke bawah oleh interpreter       Python. Hasil keluaran (*output*) yang tercetak pada layar Jupyter Notebook adalah tiga baris teks sebagai berikut:
